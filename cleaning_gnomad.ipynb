```python
import pandas as pd
import numpy as np
```

```python
cftr = pd.read_csv(r"D:\CFTR\final_cftr_dataset.csv")

gnom = pd.read_csv(r"D:\CFTR\gnomAD_v4.1.1_ENSG00000001626_2026_04_13_12_57_13.csv")
gnom.head()
```

```python
gnom.columns
```

```python
gnom = gnom[gnom["Allele Frequency"] > 0.001]
```

```python
gnom = gnom[[
    "Chromosome",
    "Position",
    "Reference",
    "Alternate",
    "Allele Frequency",
    "ClinVar Germline Classification"
]]

gnom.head()
```

```python
gnom["Chromosome"].value_counts()
```

```python
cftr.columns
```

```python
gnom["cause"] = 0
```

```python
gnom.rename(columns={"Chromosome": "chr", "Position": "pos", "Reference": "ref", "Alternate": "alt"}, inplace = True)
gnom.columns
```

```python
cftr["chr"] = cftr["chr"].astype(str)
gnom["chr"] = gnom["chr"].astype(str)

cftr["pos"] = cftr["pos"].astype(int)
gnom["pos"] = gnom["pos"].astype(int)
```

```python
final_dataset = pd.concat([cftr, gnom], ignore_index=True)

final_dataset = final_dataset.drop_duplicates(
    subset=["chr", "pos", "ref", "alt"],
    keep="first"   
)
```

```python
final_dataset = final_dataset.sample(frac=1, random_state=123).reset_index(drop=True)
```

```python
final_dataset.head()
```

```python
final_dataset.to_csv(r"D:\CFTR\final_cftr_dataset.csv", index=False)
print("Done")
```
