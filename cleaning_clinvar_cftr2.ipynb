```python
import pandas as pd
```

```python
clinvar = pd.read_excel("clinvar_result_final_tabulated.xlsx")

# Load CFTR2 (multiple sheets)
cftr2_sheets = pd.read_excel("CFTR2_30January2026.xlsx", sheet_name=None)

sheet1 = cftr2_sheets[list(cftr2_sheets.keys())[0]]
sheet2 = cftr2_sheets[list(cftr2_sheets.keys())[1]]

print(clinvar.shape, sheet1.shape, sheet2.shape)
```

```python
def clean_columns(df):
    df.columns = (
        df.columns
        .str.strip()
        .str.replace("\n", " ")
        .str.replace(r"\s+", " ", regex=True)
    )
    return df

sheet1 = clean_columns(sheet1)
sheet2 = clean_columns(sheet2)
clinvar = clean_columns(clinvar)
```

```python
cftr2 = pd.merge(
    sheet1,
    sheet2,
    on="Variant cDNA name",
    how="inner"   # ensures only matched rows
)
```

```python
clinvar["GRCh38Chromosome"] = 7
```

```python
def loc_map(location):
    start_loc = location.split(" - ")[0]
    return int(start_loc)

clinvar["pos"] = clinvar["GRCh38Location"].apply(loc_map)
```

```python
clinvar.head()
```

```python
clinvar.info()
```

```python
def spdi_ref(spdi):
    try:
        chrom, pos, ref, alt = spdi.split(":")
        return ref
    except:
        return None
    
def spdi_alt(spdi):
    try:
        chrom, pos, ref, alt = spdi.split(":")
        return alt
    except:
        return None
    
clinvar["ref"] = clinvar["Canonical SPDI"].apply(spdi_ref)
clinvar["alt"] = clinvar["Canonical SPDI"].apply(spdi_alt)
clinvar.head()
```

```python
clinvar.rename(columns = {"GRCh38Location" : "pos"}, inplace = True)
```

```python
clinvar["Germline classification"].value_counts()
```

```python
clinvar = clinvar[clinvar["Germline classification"].isin(["Pathogenic", "Benign", "Benign/Likely benign"])]
```

```python
clinvar["Germline classification"].value_counts()
```

```python
clinvar["cause"] = clinvar["Germline classification"].map({"Pathogenic" : 1, "Benign" : 0, "Benign/Likely benign": 0})
```
