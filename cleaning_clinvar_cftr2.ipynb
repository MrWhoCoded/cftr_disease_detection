```python
import pandas as pd
```

```python
clinvar = pd.read_excel("datasets/clinvar_result_final_tabulated.xlsx")

# Load CFTR2 (multiple sheets)
cftr2_sheets = pd.read_excel("datasets/CFTR2_30January2026.xlsx", sheet_name=None)

sheet1 = cftr2_sheets[list(cftr2_sheets.keys())[0]]
sheet2 = cftr2_sheets[list(cftr2_sheets.keys())[1]]

print(clinvar.shape, sheet1.shape, sheet2.shape)
```

```python
def clean_columns(df):
    df.columns = (
        df.columns
        .str.strip()
        .str.replace("\n", " ")
        .str.replace(r"\s+", " ", regex=True)
    )
    return df

sheet1 = clean_columns(sheet1)
sheet2 = clean_columns(sheet2)
clinvar = clean_columns(clinvar)
```

```python
import re

def extract_hgvs_pos(hgvs):
    match = re.search(r"g\.(\d+)", str(hgvs))
    return int(match.group(1)) if match else None

cftr2["hgvs_pos"] = cftr2["hgvs_genomic_grch38"].apply(extract_hgvs_pos)
```

```python
cftr2["pos_diff"] = cftr2["hgvs_pos"] - cftr2["pos"]

print(cftr2["pos_diff"].value_counts().head(10))
```

```python
cftr2.drop(columns=["hgvs_pos", "pos_diff"], inplace=True)
```

```python
from pyfaidx import Fasta

genome = Fasta(r"Homo_sapiens_CFTR_sequence.fa")
print(list(genome.keys()))
#print(len(genome["7"]))
```

```python
for i, row in cftr2.sample(10).iterrows():
    chrom = row["hgvs_genomic_grch38"].split(":")[0]
    pos = int(row["pos"])
    ref = str(row["ref"]).upper()

    genome_ref = str(genome[chrom][pos-1:pos-1+len(ref)]).upper()

    print("REF:    ", ref)
    print("GENOME: ", genome_ref)
    print("MATCH:  ", ref == genome_ref)
    print("-"*50)
```

```python
matches = []
failed_examples = []

for _, row in cftr2.iterrows():
    try:
        # Extract chromosome from HGVS (correct for GCF genome)
        chrom = str(row["hgvs_genomic_grch38"]).split(":")[0]
        
        pos = int(row["pos"])
        ref = str(row["ref"]).upper()

        # Get genome sequence (convert to uppercase)
        genome_ref = str(genome[chrom][pos-1:pos-1+len(ref)]).upper()

        match = (ref == genome_ref)
        matches.append(match)

        # store a few failures for debugging
        if not match and len(failed_examples) < 5:
            failed_examples.append((ref, genome_ref, chrom, pos))

    except Exception as e:
        matches.append(False)
        if len(failed_examples) < 5:
            failed_examples.append(("ERROR", str(e)))

# Results
match_rate = sum(matches) / len(matches)

print(f"Total variants: {len(matches)}")
print(f"Match rate: {match_rate:.4f}")
print(f"Matched: {sum(matches)}")
print(f"Failed: {len(matches) - sum(matches)}")

print("\nSample mismatches:")
for ex in failed_examples:
    print(ex)
```

```python
def parse_spdi(spdi):
    try:
        seq_id, pos0, ref, alt = str(spdi).split(":", 3)
        return pd.Series([seq_id, int(pos0), ref, alt])
    except:
        return pd.Series([None, None, None, None])
```

```python
clinvar[["chr", "spdi_pos0", "ref", "alt"]] = clinvar["Canonical SPDI"].apply(parse_spdi)
```

```python
matches = []
failed_examples = []

for _, row in clinvar.iterrows():
    try:
        chrom = row["chr"]          # NC_000007.14
        pos = int(row["pos"])
        ref = str(row["ref"]).upper()

        genome_ref = str(genome[chrom][pos-1:pos-1+len(ref)]).upper()
        match = (ref == genome_ref)

        matches.append(match)

        if not match and len(failed_examples) < 5:
            failed_examples.append((chrom, pos, ref, genome_ref))

    except Exception as e:
        matches.append(False)
        if len(failed_examples) < 5:
            failed_examples.append(("ERROR", str(e)))

clinvar["match"] = matches

# Stats
print("Total variants:", len(matches))
print("Match rate:", sum(matches) / len(matches))
print("Matched:", sum(matches))
print("Failed:", len(matches) - sum(matches))
print("Sample mismatches:", failed_examples)
```
